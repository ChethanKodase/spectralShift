```python
import pandas as pd

# Load the CSV

attackMode = "Lan"
#attackMode = "Vis"
df = pd.read_csv(f"/data1/chethan/spectralShift/gemma_attack/outputsStorageImagenet/backUp/alignment_shift_sweep{attackMode}.csv")



# Inspect it

print(df.head())

print(df.shape)

print(df.columns)
```

```python
point_labels = [
"query proj", "key proj", "value proj", "att output proj",
"MLP exp", "MLP out proj", "Vis-to-lan proj", "query proj",
"key proj", "value proj", "MLP gate proj", "MLP up proj",
"MLP down proj"
]



if attackMode == "Vis":
    #point_labels = point_labels[:7]
    point_labels = point_labels[:6]
else:
    #point_labels = point_labels[6:]
    point_labels = point_labels[7:]
```

```python
df['HistogramIndex'] = df['HistogramIndex'].map(dict(enumerate(point_labels)))
```
